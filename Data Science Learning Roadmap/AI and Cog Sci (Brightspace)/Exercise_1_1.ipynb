That out of the way, we'll start with a warmup, since most of you may not have programmed in a while. Below is a template for a function called `loop_summer`. In the *body of the function*, write a *loop* that adds together all of the *values* in the integer string `ints` that was passed to it. So for instance, if `loop_summer` receives the list of values

```py
[1, 1, 1, 2, 3]
```

It should *return* the *integer* `8`, since $8 = 1 + 1 + 1 + 2 + 3$.

Recall that in Python, a function may only return a value `v` if it ends with the statement `return v`, and `v` is a constant value or a variable within the [*scope* of the function](https://www.w3schools.com/PYTHON/python_scope.asp). For instance
```py
def returns_6():
    return 6
```
will always return the value 6, and
```py
def v_plus_5(v):
    in_scope = v + 5
    return in_scope
```
will return the value of `in_scope`, which is defined to be equal to `v + 5`. We could just as well have returned `v+5` without putting it in a variable.

Finally, the following function will run, but will not return anything:
```py
def does_not_return_6():
    6
```
since there is no `return` statement.

When you're done, run the *code block* with CTRL+Enter.

````python
# before you type anything, save a copy of this colab file in your google drive!
def loop_summer(ints):
    # TYPE YOUR CODE IN THIS BOX HERE
    # indented under the function definition just like this
    # hint: first, define total=0
    # then create a loop (either a while loop or for loop)
    #   inside the loop, have your code indented one over like this
    #   inside the loop you will want to add each number in ints to the total
    # we leave the loop, which is indicated by decreasing the indent
    # finally, we will want to return total
    total = 0
    for int in ints:
        total += int
    return total
````
