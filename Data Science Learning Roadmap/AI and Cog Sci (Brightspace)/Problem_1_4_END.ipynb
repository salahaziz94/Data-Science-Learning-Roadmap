We complete the lab with a breadth-first search. A breadth first search is almost identical to a depth-first search, except that it uses a queue instead of a stack, so, instead of immediately searching the current node's children, it will pop whatever is the oldest element in the queue.

This time, you will not have the same template as before, but you are allowed to copy code from the previous exercises. As inExercise 1.3, your breadth-first search function is required to return a tuple containing:

1. `True`, if the goal is found in the tree, or `False` if not
2. A list of the *labels of* the current node at each step, in order
3. A list containing lists of the child `SearchNode`s that were added to the stack, at each step

And as before, make sure to add the children of the current node *in order, from first to last*. They don't need to be in any specific order for a valid breadth-first search, but it's easier to write an evaluator that doesn't have to deal with variability.

```python
from lab1_libs.searchtree import SearchNode

def bfs(root, goal):
    # Your code goes here
    queue = [root]

    current_nodes = []
    added_nodes = []

    while queue: #while we are looking within the queue:
        current = queue.pop(0) #create a variable that pops out the first node in the queue

        current_nodes.append(current.label) #puts this in the history of what was current

        if current.label == goal:  #if the current node is the goal,
            return True, current_nodes, added_nodes #end the loop, we won! returns a tuple containing our history
        
        if current.children is not None: # checks if the current node has children, current children should equal BCD at first.
                # add the children to the queue, starting with the first child, in order
                # remember *not* to pop the children; your code will misbehave if you do!       
            for child in current.children: # a for loop that iterates over the current children list
                queue.append(child)  #put the child node in the queue (the back?)

                added_nodes.append(current.children) #adds it to the history
        else:
                added_nodes.append([]) # to keep track of our steps properly, add an empty list
        
            # If the search didn't return early, it means we failed (if it did, we'll never reach this point)
    return False, current_nodes, added_nodes

    
```
